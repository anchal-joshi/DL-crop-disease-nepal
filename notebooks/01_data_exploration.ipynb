```python
import os
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import kagglehub
```

```python

data_dir = kagglehub.dataset_download("emmarex/plantdisease")
# data_dir = Path("../data/PlantVillage")
class_counts = {}
for folder in sorted(data_dir.iterdir()):
    if folder.is_dir():
        count = len(list(folder.glob("*.jpg"))) + len(list(folder.glob("*.JPG")))
        class_counts[folder.name] = count
        print(f"{folder.name}: {count} images")
```

```python
plt.figure(figsize=(14, 6))
plt.bar(class_counts.keys(), class_counts.values(), color='steelblue')
plt.xticks(rotation=90)
plt.title("Number of images per class")
plt.tight_layout()
plt.savefig("../results/class_distribution.png")
plt.show()
```

```python
import matplotlib.image as mpimg

sample_class = "Tomato_Late_blight"
sample_folder = data_dir / sample_class
images = list(sample_folder.glob("*.jpg"))[:6]

fig, axes = plt.subplots(2, 3, figsize=(10, 6))
for ax, img_path in zip(axes.flatten(), images):
    ax.imshow(mpimg.imread(img_path))
    ax.axis("off")
fig.suptitle(f"Sample images: {sample_class}")
plt.tight_layout()
plt.savefig("../results/sample_images.png")
plt.show()
```
